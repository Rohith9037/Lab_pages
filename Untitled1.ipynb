Draw a line in a diagram from position (1, 3) to (2, 10) then to (6, 12) and finally to position (18, 20).
(Mark each point with a beautiful green colour and set line colour to red and line style dotted).

```python
import matplotlib.pyplot as plt

x = [1, 2, 6, 18]
y = [3, 10, 12, 20]

plt.plot(x,y,'r:')
plt.scatter(x, y, color="green")

plt.show()
```

```python
import matplotlib.pyplot as plt
year = [2001,2002,2003,2004,2005,2006,2007]
value = [24000,22500,19700,17500,14500,10000,5800]
plt.plot(year, value,color="red",marker="*",linestyle="--",linewidth=2,markersize=10)
plt.title("Value of Car")
plt.xlabel("Year")
plt.ylabel("Value")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt


months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]


affordable = [173, 153, 195, 147, 120, 144, 148, 109, 174, 130, 172, 131]
luxury = [189, 189, 105, 112, 173, 109, 151, 197, 174, 145, 177, 161]
super_luxury = [185, 185, 126, 134, 196, 153, 112, 133, 200, 145, 167, 110]


plt.scatter(months, affordable, color="pink", label="Affordable")
plt.scatter(months, luxury, color="yellow", label="Luxury")
plt.scatter(months, super_luxury, color="blue", label="Super Luxury")


plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Data")


plt.legend()


plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("/Iris.csv")
df
```

```python


print(df.head())
```

```python

#b
plt.scatter(df['SepalLengthCm'], df['SepalWidthCm'])

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

plt.show()
```

```python
#c
```

```python
#d
plt.plot(df["SepalLengthCm"], df["SepalWidthCm"])

plt.title("Line Plot")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

plt.show()
```

```python
#e
plt.hist(df["SepalLengthCm"], bins=10)

plt.title("Histogram of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")

plt.show()
```

```python
#f
total = df.groupby("variety")["SepalLengthCm"].sum()

plt.bar(total.index, total.values)

plt.title("Total Sepal Length")
plt.xlabel("Variety")
plt.ylabel("Total Sepal Length")

plt.show()
```
