```python
import numpy as np

arr1 = np.array(list(map(int, input("Enter first array elements: ").split())))
arr2 = np.array(list(map(int, input("Enter second array elements: ").split())))

print("Greater:", np.greater(arr1, arr2))
print("Greater or Equal:", np.greater_equal(arr1, arr2))
print("Lesser:", np.less(arr1, arr2))
```
