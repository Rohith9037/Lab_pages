```python
import matplotlib.pyplot as plt
```

```python
monthly = ["jan", "feb", "mar", "apr", "may"]
sales1 = [100, 150, 120, 130, 180]
sales2 = [180, 130, 160, 170, 100]

plt.plot(monthly,sales1,label="Product A", color='black', marker='o', linestyle='dashed',linewidth=1, markersize=5)
plt.plot(monthly,sales2,label="Product B", color='green', marker='o', linestyle='dashed',linewidth=1, markersize=5)

plt.title("monthly sales")
plt.xlabel("Month")
plt.ylabel("sales")
plt.grid()
plt.tight_layout()
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

#a
df = pd.read_csv('Iris.csv')
df.head(5)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['SepalLengthCm'],df['SepalWidthCm'], color='blue', s=80)
plt.title("Iris data")
plt.xlabel("SepalLengthCm")
plt.ylabel("SepalWidthCm")
plt.grid(linestyle= '-',alpha=0.5)
plt.show()
```

```python
#c
plt.figure(figsize=(10, 6))
for species in df['Species'].unique():
    subset = df[df['Species'] == species]
    plt.scatter(
        subset['SepalLengthCm'],
        subset['SepalWidthCm'],
        label=species
    )
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Iris Dataset')
plt.legend()
plt.show()
```

```python
#d
plt.figure(figsize=(10, 6))
plt.plot(df['SepalLengthCm'], df['SepalWidthCm'])
plt.title("Iris data")
plt.xlabel("SepalLengthCm")
plt.ylabel("SepalWidthCm")
plt.grid(linestyle= '-',alpha=0.5)
plt.show()
```

```python
#e
plt.figure(figsize=(10, 6))
plt.hist(df['PetalLengthCm'], bins=10)
plt.title("Iris data")
plt.xlabel("PetalLengthCm")
plt.ylabel("Frequency")
plt.grid(linestyle= '-',alpha=0.5)
plt.show()
```

```python
#f
plt.figure(figsize=(10, 6))
df.groupby('Species')['SepalLengthCm'].sum().plot(kind='bar')
plt.title("Iris Data")
plt.xlabel("Species")
plt.ylabel("Total Sepal Length")
plt.grid(linestyle='-', alpha=0.5)
plt.show()
```

```python
#g
plt.figure(figsize=(10, 6))
df.boxplot(column='SepalLengthCm', by='Species')
plt.title('Distribution of Sepal Length by Iris Variety')
plt.suptitle('')
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.grid()
plt.show()
```

```python
#h
plt.figure(figsize=(8, 8))
species_count = df['Species'].value_counts()
plt.pie(
    species_count,
    labels=species_count.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Percentage Distribution of Iris Varieties')
plt.show()
```

```python
#i
plt.figure(figsize=(10, 6))

for variety in df['Species'].unique():
    data = df[df['Species'] == variety]

    plt.scatter(
        data['PetalLengthCm'],
        data['PetalWidthCm'],
        label=variety
    )

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Length vs Petal Width")
plt.legend()
plt.grid(alpha=0.5)

plt.show()
```

```python
#J
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].hist(df['SepalLengthCm'], bins=10, edgecolor='black')
axes[0, 0].set_title("Histogram of Sepal Length")

axes[0, 1].hist(df['SepalWidthCm'], bins=10, edgecolor='black')
axes[0, 1].set_title("Histogram of Sepal Width")

axes[1, 0].hist(df['PetalLengthCm'], bins=10, edgecolor='black')
axes[1, 0].set_title("Histogram of Petal Length")

axes[1, 1].hist(df['PetalWidthCm'], bins=10, edgecolor='black')
axes[1, 1].set_title("Histogram of Petal Width")

plt.tight_layout()
plt.show()
```

```python
#K
data = df.groupby('Species')[['SepalLengthCm', 'SepalWidthCm']].mean()

data.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title("Average Sepal Length and Sepal Width")
plt.xlabel("Iris Variety")
plt.ylabel("Average")
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.5)
plt.show()
```

```python
#L
plt.figure(figsize=(10, 6))
for Species in df['Species'].unique():
    data = df[df['Species'] == Species]

    plt.scatter(
        data['PetalLengthCm'],
        data['PetalWidthCm'],
        s=data['SepalLengthCm'] * 50,
        alpha=0.6,
        label=variety
    )

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Bubble Scatter Plot")
plt.legend()
plt.grid(alpha=0.5)
plt.show()
```
