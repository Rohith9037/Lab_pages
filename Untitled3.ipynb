```python
text="""
Django is a high-level, open-source Python web framework designed to help developers build secure, scalable, and maintainable websites quickly.
 It follows the Model-Template-View (MTV) architectural pattern and adheres to the "Don't Repeat Yourself" (DRY) principle, which minimizes code duplication.
  Built by experienced developers, Django handles much of the hassle of web development with its famous "batteries-included" philosophy, offering built-in tools for user authentication, database management, and administrative interfaces right out of the box. Its strong emphasis on built-in security helps developers avoid common mistakes like SQL injection,
 cross-site scripting, and csrf attacks, making it a trusted platform for everything from small startups to large-scale organizations like Instagram, Pinterest, and Disqus."""
with open("django_notes.txt", "w") as file:
  file.write(text)
```

```python
with open("django_notes.txt", "r") as file:
  text=file.read()
```

```python
def create_chunks(text,chunk_size=200):
  return[
      text[i:i + chunk_size]
      for i in range(0,len(text),chunk_size)
  ]

chunks = create_chunks(text)

for i,chunk in enumerate(chunks):
  print(f"Chunk {i}:")
  print(chunk)
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-miniLM-L6-v2")
embeddings = model.encode(chunks)
print(embeddings.shape)
```

```python
import faiss
import numpy as np
embeddings = np.array(embeddings).astype("float32")
diamension = embeddings.shape[1]
index = faiss.IndexFlatL2(diamension)
index.add(embeddings)
print("Number of vectors:",index.ntotal)
```

```python
question="what is urls.py used for?"
```

```python
question_embedding=model.encode([question])
question_embedding=np.array(question_embedding).astype("float32")
distance,indices=index.search(
    question_embedding,
    k=3
)
for i in indices[0]:
  print("\n---")
  print(chunks[i])
```

```python
from openai import OpenAI
from google.colab import userdata
groq_api_key = userdata.get("subi")

client = OpenAI(
    api_key=groq_api_key,
    base_url="https://api.groq.com/openai/v1"
)
context = "\n\n".join(
    chunks[i] for i in indices[0]
)
messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant that answers question based on the provided context."
    },
    {
        "role": "user",
        "content": f"""
Answer the question using only the context brlow.

context:
{context}
Question:
{question}
if the answer is not in the context,
say "I don't know based on the provided document."
"""
    }
]

response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=messages
)
print(response.choices[0].message.content)
```
